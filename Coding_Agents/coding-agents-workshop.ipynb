## 1. What are coding agents? *(~10 mins)*

### From Autocomplete to Autonomous Systems

Traditional developer tooling and early AI helpers operated as inline text generators. Coding agents represent a shift from assistive prediction to goal-driven execution:

* **Tab Autocomplete (Copilot / Ghost text):** Predicts the next few tokens based on surrounding context. Fast, but limited to immediate syntax completion.
* **Chat Assistants:** Responds to isolated prompts in a side panel. Can explain code, write snippets, or answer queries, but requires you to manually copy/paste and test the code yourself.
* **Autonomous Coding Agents:** Systems given a high-level goal in natural language. They reason through the problem, inspect the project environment, generate and apply diffs across multiple files, execute shell commands (running linters, compilers, tests), and self-correct when errors occur.

### Core Traits of an Agentic System

* **Goal Orientation:** Works toward a functional milestone (e.g., *"Add authentication with OAuth and write test cases"*) rather than just writing a single function.
* **Context Awareness:** Automatically navigates repository structure, dependencies, and git history rather than relying on manually pasted snippets.
* **Tool Calling & Sandboxed Execution:** Can read files, edit code, execute terminal commands, run browser tests, and call external services.
* **Observability & Feedback Loops:** Detects compiler/test failures, analyzes trace logs, and adjusts its approach without human intervention for every error.

## 3. How do coding agents work? *(~10 mins)*

Understanding how agents think prevents them from going off the rails:

```
[ User Prompt / Goal ]
          │
          ▼
┌──────────────────┐
│  Reason & Plan   │ ◄────────────────────────┐
└─────────┬────────┘                          │
          │                                   │
          ▼                                   │
┌──────────────────┐     Tool Outputs / Logs  │ (Self-Correction Loop)
│   Take Action    │ ─────────────────────────┘
│  (Read/Edit/Run) │
└─────────┬────────┘
          │ (Goal Verified)
          ▼
[ Final Diff / PR ]

```

### The Architecture Under the Hood

1. **The ReAct Pattern (Reasoning + Acting):** Agents maintain a continuous loop:
    * **Thought:** Assesses current state and plans the next atomic step.
    * **Action:** Issues a structured tool call (e.g., `readFile("src/api.ts")` or `runTerminal("npm test")`).
    * **Observation:** Evaluates stdout/stderr from the tool execution.


2. **Context Window & Retrieval:**
    * Uses lexical search (grep/ripgrep), semantic embeddings, and abstract syntax tree (AST) indexing to pull relevant files into context without overflowing token limits.


3. **Artifacts & State Tracking:**
    * Modern agents maintain structured state: task breakdowns, scratchpads, implementation checklists, and reviewable git diffs.


4. **Verification Step:**
    * High-performing agents verify their own work before reporting completion by running unit tests, type checkers, and linters.

## 5. Tips, Tricks & Advanced Workflows *(~15 mins)*

### A. Model Context Protocol (MCP) Servers

* **What is MCP?** An open standard that connects agents to external tools, databases, APIs, and local resources without hardcoding integrations.
* **Why use MCP?**
    * **Database Inspection:** Connect an MCP server to PostgreSQL/SQLite so the agent inspects real schemas instead of guessing table columns.
    * **Browser & Docs:** Add a documentation search MCP (e.g., PyTorch docs, Next.js docs) so the agent references up-to-date APIs rather than stale training data.
    * **Issue Trackers:** Allow agents to pull context directly from GitHub issues, Jira, or Linear.



### B. Custom Agent Skills & Workspace Rules

* **Project Guidelines (`.antigravity/rules` or project instructions):**
    * Define repository conventions: strict TypeScript types, preferred test frameworks (e.g., `vitest` over `jest`), styling rules, and architectural boundaries.


* **Custom Skills:**
    * Define reusable prompt-and-tool chains for common actions (e.g., *"Generate DB Migration"*, *"Benchmark Service"*, *"Check Security Vulnerabilities"*).



### C. Prompt Steering & Workflow Best Practices

1. **Provide Clear Acceptance Criteria:** Avoid vague instructions like *"make this faster."* Instead, say: *"Refactor `calculateStats` to use vector operations. Add benchmarks in `test/bench.py` and ensure elapsed time drops below 50ms."*
2. **Scaffold Before Delegating:** Give the agent an existing pattern to follow (e.g., an existing test file or service interface) to keep code style consistent.
3. **Never Blindly Accept Diffs:** Always review the agent's changes in git diff view—check for unintentional deletions, hallucinated dependencies, or disabled type checks.
4. **Use Atomic Commits:** Ask the agent to work on one self-contained feature or bug fix at a time so rollbacks remain trivial.
